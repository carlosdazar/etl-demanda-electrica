# Restructuración
Trasformación del formato largo al formato ancho, se debe pivotear
las categorías (fuentes, sector, indicadores) en columnas independientes.

Union de los 3 csv

```python
# Indicador para identificar de donde viene cada variable
df1_trasform['Indicador'] = 'Generacion'
df2_trasform['Indicador'] = 'Capacidad'
df3_trasform['Indicador'] = 'Consumo'
```

```python
# union 3 csv
df_final = pd.concat([df1_trasform, df2_trasform, df3_trasform], ignore_index=True)
# Unificar fuente y Sector en una sola columna
df_final['Variable'] = df_final['Fuente'].fillna(df_final['Sector'])
# Eliminar  filas "Total" para evitar duplicar datos de los componentes
df_final_sin = df_final[df_final['Variable'] != 'Total'].copy()
```

```python
#pivot de columnas
df_pivot = df_final_sin.pivot_table(
    index=['País', 'Año'],
    columns=['Indicador', 'Variable'],
    values='Valor',
    aggfunc='sum'
).reset_index()
# aplanar nombre de columnas
df_pivot.columns = ['_'.join(col).strip('_') for col in df_pivot.columns.values]
```

```python
#comprobar si existen valores nulos
print("Nulos restantes en todo el DataFrame:", df_pivot.isna().sum().sum())
```

Con el fin de estandarizar el dataset para su posterior merge y carga se aplica lo siguiente

```python
# Normalizar acentos y quitar tildes/caracteres no ASCII
df_pivot.columns = df_pivot.columns.str.normalize('NFKD').str.encode('ascii', errors='ignore').str.decode('utf-8')
# Remplazar comas por  '_'
df_pivot.columns = df_pivot.columns.str.replace(r',\s*', '-', regex=True)
# Reemplazar caracteres especiales y espacios por '_' y pasar a minúsculas
df_pivot.columns = (
    df_pivot.columns
    .str.replace(r'[^0-9a-zA-Z]+', '_', regex=True)
    .str.strip('_')
    .str.lower()
)
# Renombrar la columna del año (como 'año' -> 'ano' -> 'anio')
df_pivot.rename(columns={'ano': 'anio'}, inplace=True)
```

```python
# Normalizar tildes en la columna 'pais'
for d in [df_pivot]:
    d['pais'] = (
        d['pais']
        .astype(str)
        .str.normalize('NFKD')
        .str.encode('ascii', errors='ignore')
        .str.decode('utf-8')
        .str.strip()
        .str.lower()
    )
df=df_pivot.copy()
```
